Figures 1–8 are built by `evoproto.figures`, and the ones that show data call
the same code paths as the protocol, so a figure cannot drift away from the
numbers it illustrates.


```python
# Google Colab: install the package from the repository (skip if running locally)
import importlib.util, subprocess, sys

if importlib.util.find_spec('evoproto') is None:
    IN_COLAB = 'google.colab' in sys.modules
    if IN_COLAB:
        subprocess.run(['pip', 'install', '-q',
                        'git+https://github.com/omega2417/bnt.git#subdirectory=evoproto'],
                       check=True)
    else:
        sys.path.insert(0, '../src')   # running from notebooks/ in a clone

import evoproto
print('evoproto', evoproto.__version__)
```

```python
from evoproto.figures import FIGURES, render_all, set_style
set_style()
print('builders:', sorted(FIGURES))
```

## Render all eight figures

PDF (vector, for the publisher, with fonts embedded) and 600 dpi PNG.


```python
paths = render_all('figures')
for path in paths:
    print(path)
```

## Inspect them inline


```python
from IPython.display import Image, display

for name in sorted(FIGURES):
    print(name)
    display(Image(filename=f'figures/{name}.png', width=900))
```

## Rebuild a single figure with different parameters

Figure 7 runs the dry run itself, so it takes the protocol's knobs.


```python
from evoproto.figures import fig_7

figure = fig_7(replicates=4, population=24, generations=15)
figure.savefig('figures/fig7_small.png', dpi=150)
display(Image(filename='figures/fig7_small.png', width=900))
```
