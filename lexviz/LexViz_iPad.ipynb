Виконуйте клітинки по черзі (Shift+Enter). Ноутбук має лежати в корені розпакованої папки `lexviz`.

```python
import os, sys
sys.path.insert(0, os.getcwd())
from lexviz import load, build
from lexviz.figures import FIGURES
from lexviz.palette import report

d = load('data/ukraine_recovery_2026.yaml')   # валідатор спрацює тут
print('OK:', len(d['areas']), 'сфер; правила розділу 2.3 виконано')
print(report()[1])
```

## Перелік типів рисунків

```python
for i, (name, m) in enumerate(FIGURES.items(), 1):
    print(i, name, '-', m.TITLE)
```

## Побудувати один рисунок і показати його тут

```python
%matplotlib inline
fig = FIGURES['fig4_compliance_matrix'].draw(d)
fig.set_dpi(110)
fig
```

## Перевірка правила: змініть оцінку й подивіться, як валідатор зупинить збірку

```python
import copy
from lexviz import validate, DataError
bad = copy.deepcopy(d)
next(a for a in bad['areas'] if a['id'] == 'state_aid')['authors']['rating'] = 'partly_aligned'
try:
    validate(bad)
except DataError as e:
    print('Валідатор:', e)
```

## Усі рисунки для видавця (PNG 600 dpi, PDF, SVG) + галерея output/index.html

```python
out = build(d, 'output', dpi=600)
print(sum(len(v) for v in out.values()), 'файлів')
```
